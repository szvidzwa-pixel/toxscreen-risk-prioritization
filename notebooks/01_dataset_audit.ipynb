# Dataset Audit

Use this notebook to inspect the ClinTox file before running the training pipeline.

The goal of the EDA is to make the project story visible before modeling:

- we use molecular structure data to predict clinical toxicity risk early
- the `CT_TOX` target is the endpoint of interest
- false negatives matter because missing a toxic compound is costly


```python
from src.toxscreen.data import load_dataset, profile_dataset

df = load_dataset('data/raw/clintox.csv')
df.head()
```

```python
profile_dataset(df)
```

```python
df.isna().sum()
```

```python
df['CT_TOX'] = df['ct_tox'] if 'ct_tox' in df.columns else df['CT_TOX']
df['FDA_APPROVED'] = df['fda_approved'] if 'fda_approved' in df.columns else df['FDA_APPROVED']

df['CT_TOX'].value_counts()
```

```python
df['CT_TOX'].value_counts(normalize=True)
```

```python
df['FDA_APPROVED'].value_counts(normalize=True)
```

```python
df['smiles'].str.len().describe()
```

```python
import matplotlib.pyplot as plt

ct_norm = df['CT_TOX'].value_counts(normalize=True).sort_index()
ct_norm.plot(kind='bar', color=['#7aa6c2', '#d95f5f'], title='ClinTox Target Distribution')
plt.ylabel('Proportion')
plt.show()
```
